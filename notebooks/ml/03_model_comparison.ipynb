# Fraud Detection — Model Comparison

This notebook extends the fraud detection analysis by comparing the Logistic Regression baseline with alternative classification models.

The main goals are to:
- establish Logistic Regression as the baseline model,
- train an alternative model on the same training data,
- evaluate both models using the same validation dataset and metrics,
- determine whether a different modeling approach can better distinguish fraudulent from non-fraudulent transactions.

The models are evaluated primarily using recall, precision and F1-score, as accuracy is not informative enough for the highly imbalanced fraud detection problem.
