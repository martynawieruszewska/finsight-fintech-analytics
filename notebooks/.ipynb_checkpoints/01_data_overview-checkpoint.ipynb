## Codes dataset

```python
with open("../data/raw/mcc_codes.json", "r") as file:
    mcc_dict = json.load(file)
type(mcc_dict) 
list(mcc_dict.items())[:5]
codes = pd.DataFrame(
    mcc_dict.items(),
    columns=["mcc", "description"]
)
codes.head()
```

```python
codes.info()
```

```python
codes.dtypes
```

```python
codes.describe()
```

## Fraud labels dataset

```python
with open("../data/raw/train_fraud_labels.json", "r") as file:
    fraud_dict = json.load(file)
type(fraud_dict)
list(fraud_dict.items())[:5]
fraud_target = fraud_dict["target"]
type(fraud_target)
fraud = pd.DataFrame(
    fraud_target.items(),
    columns=["transaction_id", "is_fraud"]
)
fraud.head()
```

```python
fraud.shape
```

```python
fraud["is_fraud"].value_counts()
```

```python
fraud.info()
```
